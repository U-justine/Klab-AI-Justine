```python
class SimpleClassifier:
    def __init__(self, threshold=0.5):
        self.threshold= threshold
        self.is_ifiltered= False
        
    def fit(self, scores, labels):
        positives = [s for s , y in zip(scores,labels) if y==1]
        self.threshold = sum(positives) / len(positives)
        self.is_ifiltered= True
        return self
    
    def predict(self, scores):
        if not self.is_ifiltered:
            raise RuntimeError("call fit() before Predict")
        return [1 if s >= self.threshold else 0 for s in scores]
    
model= SimpleClassifier().fit([0.2, 0,4, 0.7, 0.9], [0,0,1,1])

print (model.threshold, model.predict([0.3,0.85]))
        
    
    
```
